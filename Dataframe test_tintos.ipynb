```python
import pandas as pd
from IPython.display import display
```

```python
df1 = pd.read_csv('C:\\Users\\Pauline\\Downloads\\Proyecto_grupo2_vinos\\tintos\\DF_BF_tintos_150to202.csv')
df2 = pd.read_csv('C:\\Users\\Pauline\\Downloads\\Proyecto_grupo2_vinos\\tintos\\csv_tintos_1_a_15000_BeautifulSoup.csv', skiprows=1)
```

```python
print(df1.shape)
print(df2.shape)

print(df1.columns)
print(df2.columns)
```

```python
# Eliminar las filas vacías
df_cleaned = df2.dropna(how='all')  # Elimina filas donde todos los valores son NaN

# Eliminar las filas donde aparece el título '---' en la columna 'Url'
# Esto eliminará las filas que contienen texto como "--- wine_data_batch_38.csv ---"
df_cleaned = df_cleaned[~df_cleaned['Url'].str.contains('---', na=False)]

# Verifica los resultados
print(f"DataFrame limpio, filas restantes: {df_cleaned.shape[0]}")

# Guardar el DataFrame limpio en un nuevo archivo CSV
df_cleaned.to_csv(r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\tintos\BF_tintos_1to150_clean', index=False)

df2 = df_cleaned
```

```python
# Concatenar los DataFrames de BF
df_tintos_BF_final = pd.concat([df1, df2], ignore_index=True)
#Eliminamos duplicated por ID
df_tintos_BF_final = df_tintos_BF_final.drop_duplicates(subset=['ID'], keep='first')

# Verifica los resultados
print(f"DataFrame combinado tiene {df_tintos_BF_final.shape[0]} filas y {df_tintos_BF_final.shape[1]} columnas.")

# Guardar el DataFrame combinado en un nuevo archivo CSV
df_tintos_BF_final.to_csv(r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\tintos\1df_tintos_BF_final.csv', index=False)
```

```python
#Igual para caracteristicas : 

df3 = pd.read_csv('C:\\Users\\Pauline\\Downloads\\Proyecto_grupo2_vinos\\tintos\\caracteristicas_tintos_150to202.csv')
df4 = pd.read_csv('C:\\Users\\Pauline\\Downloads\\Proyecto_grupo2_vinos\\tintos\\csv_tintos_1_a_15000_características.csv')
```

```python
print(df3.shape)
print(df4.shape)

print(df3.columns)
print(df4.columns)
```

```python
# Concatenar los DataFrames de BF
df_tintos_caracteristicas_final = pd.concat([df3, df4], ignore_index=True)
#Eliminamos duplicated x ID
df_tintos_caracteristicas_final = df_tintos_caracteristicas_final.drop_duplicates(subset=['ID'], keep='last')


# Verifica los resultados
print(f"DataFrame combinado tiene {df_tintos_caracteristicas_final.shape[0]} filas y {df_tintos_caracteristicas_final.shape[1]} columnas.")

# Guardar el DataFrame combinado en un nuevo archivo CSV
df_tintos_caracteristicas_final.to_csv(r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\tintos\1df_tintos_caracteristicas_final.csv', index=False)
```

```python
#Contar el número de duplicados en la columna 'ID'

duplicados_bf = df_tintos_BF_final['ID'].duplicated().sum()
print(f"Cantidad de duplicados en la columna 'ID' de Beautiful Soup: {duplicados_bf}")

#Mostrar las filas duplicadas por 'ID' ordenadas por 'ID'
duplicados_filas_bf = df_tintos_BF_final[df_tintos_BF_final['ID'].duplicated(keep=False)].sort_values(by='ID')  # keep=False mantiene todas las filas duplicadas
print(duplicados_filas_bf)


#Igual con caracteristicas :

duplicados_car = df_tintos_caracteristicas_final['ID'].duplicated().sum()
print(f"Cantidad de duplicados en la columna 'ID' de DF de caracteristicas: {duplicados_car}")

#Mostrar las filas duplicadas por 'ID' ordenadas por 'ID'
duplicados_filas_car = df_tintos_caracteristicas_final[df_tintos_caracteristicas_final['ID'].duplicated(keep=False)].sort_values(by='ID')  # keep=False mantiene todas las filas duplicadas
print(duplicados_filas_car)



```

```python
#Eliminar espacios blancos : 
df_tintos_BF_final['ID'] = df_tintos_BF_final['ID'].astype(str).str.strip()
df_tintos_caracteristicas_final['ID'] = df_tintos_caracteristicas_final['ID'].astype(str).str.strip()
```

```python
# Encontrar la intersección de los IDs
ids_comunes_BF_car = pd.merge(df_tintos_BF_final[['ID']], df_tintos_caracteristicas_final[['ID']], on='ID', how='inner')

# Contar cuántos IDs están en común
cantidad_ids_comunes_BF_car = ids_comunes_BF_car.shape[0]
print(cantidad_ids_comunes)
```

```python
#MERGED LOS 2 ARCHIVOS : BF + CRACTERISTICAS

# Si un ID del df no tiene info en los otros DataFrames, se rellenará con "No disponible"
merged_df_tintos = pd.merge(df_tintos_BF_final, df_tintos_caracteristicas_final, on="ID", how="left")

# Rellenar los valores NaN con "No disponible"
merged_df_tintos.fillna("No disponible", inplace=True)

#Eliminamos duplicated x ID
merged_df_tintos = merged_df_tintos.drop_duplicates(subset=['ID'], keep='last')

# Guardar el DataFrame final como un nuevo CSV
final_df_path_tintos = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\tintos\0_Final_DF_BF+caracteristicas_tintos.csv'
merged_df_tintos.to_csv(final_df_path_tintos, index=False)
print(f"Datos guardados en {final_df_path_tintos}")
print(merged_df_tintos.shape)
```
