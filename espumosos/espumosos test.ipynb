```python
import pandas as pd 
```

```python
df = pd.read_csv('df_espumosos_ok2.csv')
```

```python
df.head
```

```python
#Lista de columnas a verificar (excluyendo 'url' e 'ID')
columns_to_check = [ 'Año', 'País', 'Región', 'Bodega', 'Tipo de vino']

# Filtrar filas donde todas las columnas tienen "No disponible"
df_missing = df[df[columns_to_check].eq('No disponible').all(axis=1)]

# Mostrar el DataFrame con las filas que tienen "No disponible" en todas las columnas
print("\nFilas con 'No disponible' en todas las columnas:")
print(df_missing)

# Eliminar esas filas del DataFrame original
df = df[~df.index.isin(df_missing.index)]

# Mostrar el DataFrame original después de eliminar las filas
print("\nDataFrame Original después de eliminar las filas con 'No disponible':")
print(df)
```

```python
df_missing.shape
```

```python
df.shape
```

```python
# Guardar df en un archivo CSV
df.to_csv('df_espumosos_ok.csv', index=False)

# Guardar df_missing en un archivo de texto
df_missing['Url'].to_csv('df_missing.txt', index=False, header=True, sep='\t')
```

```python
# Cargar ambos archivos CSV en DataFrames
df1 = pd.read_csv('df_espumosos_ok.csv')  # Cambia por la ruta de tu primer archivo
df2 = pd.read_csv('df_espumosos_ok2.csv')  # Cambia por la ruta de tu segundo archivo

# Concatenar los DataFrames
df_combined = pd.concat([df1, df2], ignore_index=True)

# Guardar el DataFrame combinado en un nuevo archivo CSV
df_combined.to_csv('all_espumosos_ok.csv', index=False)

print("Archivos combinados y guardados correctamente como 'archivo_combinado.csv'")
```
