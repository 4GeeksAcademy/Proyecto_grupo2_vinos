```python
import pandas as pd

# Ruta de los archivos CSV
df_path = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\DF_BF_all_espumosos_ok.csv'
caracteristicas_path = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\caracteristicas_espumosos.csv'
notas_path = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\notas_de_sabor_limpio.csv'

# Leer los CSVs
df = pd.read_csv(df_path)
caracteristicas = pd.read_csv(caracteristicas_path)
notas = pd.read_csv(notas_path)

# Verificar si hay duplicados en la columna 'ID' de ambos DataFrames
print(df['ID'].duplicated().sum())
print(caracteristicas['ID'].duplicated().sum())
print(notas['ID'].duplicated().sum())


```

```python
# Eliminar duplicados en la columna 'ID' de cada DataFrame
df = df.drop_duplicates(subset='ID')
caracteristicas = caracteristicas.drop_duplicates(subset='ID')
notas = notas.drop_duplicates(subset='ID')

# Verificar nuevamente si hay duplicados después de eliminarlos
print("\nDuplicados después de eliminar:")
print("Duplicados en df['ID']:", df['ID'].duplicated().sum())
print("Duplicados en caracteristicas['ID']:", caracteristicas['ID'].duplicated().sum())
print("Duplicados en notas['ID']:", notas['ID'].duplicated().sum())
```

```python
df_BF = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\DF_BF_not_clean.csv'
df_BF = pd.read_csv(df_BF)
print("Duplicados en df['ID']:", df_BF['ID'].duplicated().sum())

df_BF = df_BF.drop_duplicates(subset='ID')
print("Duplicados en df['ID']:", df_BF['ID'].duplicated().sum())

df_BF.to_csv(r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\DF_BF_all_espumosos_ok.csv', index=False)
```

```python
# Comparar 'df' con 'caracteristicas' para encontrar los IDs que faltan en 'caracteristicas'
missing_in_caracteristicas = df[~df['ID'].isin(caracteristicas['ID'])]

# Comparar 'df' con 'notas' para encontrar los IDs que faltan en 'notas'
missing_in_notas = df[~df['ID'].isin(notas['ID'])]

# Mostrar el resultado de los datos faltantes
print("Filas con IDs que faltan en 'caracteristicas':")
print(missing_in_caracteristicas.shape)

print("Filas con IDs que faltan en 'notas de sabor':")
print(missing_in_notas.shape)
```

```python
# Seleccionar solo la columna 'Url' del DataFrame 'missing_in_notas'
missing_in_notas_url = missing_in_notas[['Url']]

# Guardar el DataFrame con solo la columna 'Url' en el archivo CSV
missing_in_notas_url.to_csv(r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\notas_de_sabor_faltantes.csv', index=False)
```

```python
#MERGED 2 ARCHIVOS

# Ruta de los archivos CSV
df_path = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\DF_BF_all_espumosos_ok.csv'
caracteristicas_path = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\caracteristicas_espumosos.csv'

# Leer los CSVs
df = pd.read_csv(df_path)
caracteristicas = pd.read_csv(caracteristicas_path)

# Realizar el merge, uniendo los tres DataFrames por la columna 'ID' (de forma externa para incluir todos los registros)
# Si un ID del df no tiene info en los otros DataFrames, se rellenará con "No disponible"
merged_df = pd.merge(df, caracteristicas, on="ID", how="left")

# Rellenar los valores NaN con "No disponible"
merged_df.fillna("No disponible", inplace=True)

# Guardar el DataFrame final como un nuevo CSV
final_df_path = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\DF_BF+caracteristicas_espumosos_.csv'
merged_df.to_csv(final_df_path, index=False)
print(f"Datos guardados en {final_df_path}")
```
