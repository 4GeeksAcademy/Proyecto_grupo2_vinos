```python
import os
import pandas as pd

# Ruta de la carpeta donde están los CSV
folder_path = r'C:\Users\Pauline\Downloads\Proyecto_grupo2_vinos\espumosos\notas de sabor'

# Lista para almacenar los DataFrames de cada archivo CSV
all_csv_files = []

# Iterar sobre todos los archivos en la carpeta
for filename in os.listdir(folder_path):
    # Comprobar si el archivo es un archivo CSV
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)
        
        # Leer el CSV y añadirlo a la lista
        df = pd.read_csv(file_path)
        all_csv_files.append(df)

# Concatenar todos los DataFrames en uno solo
combined_df_ns_espu = pd.concat(all_csv_files, ignore_index=True)


#print(combined_df['ID'].duplicated().sum())

combined_df_ns_espu = combined_df_ns_espu.drop_duplicates(subset='Wine_ID', keep='first')

print(combined_df_ns_espu['Wine_ID'].duplicated().sum())

combined_df_ns_espu.rename(columns={'Wine_ID': 'ID'}, inplace=True)

# Verificar las primeras filas del DataFrame combinado
print(combined_df_ns_espu.head())
print(combined_df_ns_espu.shape)


combined_df_ns_espu.to_csv(r'C:\\Users\\Pauline\\Downloads\\Proyecto_grupo2_vinos\\espumosos\\DF_NS_total_espu.csv', index=False)
```
