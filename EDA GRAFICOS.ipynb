```python
#Creamos df UVA: 
df_uva = df_tintos[['ID', 'Valoración', 'Uva']].copy()

# Separamos palabras de la columna uva: 
df_uva_separado = df_uva['Uva'].str.split(',', expand=True)

# Paso 2: Renombrar las nuevas columnas con nombres únicos, por ejemplo, 'Uva_1', 'Uva_2', etc.
df_uva_separado.columns = [f'Uva_{i+1}' for i in range(df_uva_separado.shape[1])]

# Paso 3: Concatenar las columnas separadas de 'Uva' con 'ID' y 'Valoración'
df_uva = pd.concat([df_uva[['ID', 'Valoración']], df_uva_separado], axis=1)

# Paso 4: Si hay valores nulos, los reemplazamos por "No disponible"
df_uva = df_uva.fillna('No disponible')
df_uva['Valoración'] = pd.to_numeric(df_uva['Valoración'], errors='coerce')



# Mostrar el resultado final
#print(df_uva.shape)
#print(df_uva.head())
#print(df_uva.columns)
```

```python

#Todas las columnas de 'Uva_1' a 'Uva_16' en una sola lista
uvas = df_uva[['Uva_1', 'Uva_2', 'Uva_3', 'Uva_4', 'Uva_5', 'Uva_6', 'Uva_7', 'Uva_8', 'Uva_9', 'Uva_10', 'Uva_11', 'Uva_12', 'Uva_13', 'Uva_14', 'Uva_15', 'Uva_16']].values.flatten()

# Convertir la lista de uvas a un DataFrame para contar las frecuencias
uvas_series = pd.Series(uvas)
uvas_series = uvas_series[uvas_series != 'No disponible']
frecuencia_uvas = uvas_series.value_counts()

# Paso 3: Calcular el promedio de la 'Valoración' para cada uva
# Primero, creamos un diccionario donde cada uva tiene su correspondiente promedio de valoración.
# Usamos un diccionario para acumular las valoraciones y contar cuántas veces aparece cada uva.

promedios = {}
for uva in frecuencia_uvas.index:
    # Filtrar las filas donde la uva está presente
    filas_uva = df_uva[df_uva[['Uva_1', 'Uva_2', 'Uva_3', 'Uva_4', 'Uva_5', 'Uva_6', 'Uva_7', 'Uva_8', 'Uva_9', 'Uva_10', 'Uva_11', 'Uva_12', 'Uva_13', 'Uva_14', 'Uva_15', 'Uva_16']].isin([uva]).any(axis=1)]
    
    # Calcular el promedio de la valoración para esas filas
    promedio_valoracion = filas_uva['Valoración'].mean()
    
    # Guardar en el diccionario
    promedios[uva] = promedio_valoracion

#DF con la frecuencia y el promedio de valoración
df_resultado = pd.DataFrame({
    'Frecuencia': frecuencia_uvas,
    'Promedio_Valoración': [promedios[uva] for uva in frecuencia_uvas.index]
})

df_top5 = df_resultado.head(5)

# Paso 6: Graficar el análisis multivariante
plt.figure(figsize=(12, 6))

# Graficar la frecuencia de cada uva en un gráfico de barras
sns.barplot(x=df_top5.index, y=df_top5['Frecuencia'], color='indianred', label='Frecuencia')

# Graficar el promedio de la valoración en el mismo gráfico (con un eje y secundario)
ax2 = plt.gca().twinx()
sns.lineplot(x=df_top10.index, y=df_top5['Promedio_Valoración'], marker='o', color='black', ax=ax2, label='Promedio Valoración')

# Títulos y etiquetas
plt.title('Top 5 Uvas: Frecuencia y Promedio de Valoración',  fontweight='bold', fontsize=16, pad=20)
plt.xlabel('Uvas')
plt.ylabel('Frecuencia')
ax2.set_ylabel('Promedio de Valoración')

ax2.legend(loc='upper right')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
